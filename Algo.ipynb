```python
#longest palindrome
def longest_palindrome_bf(s):
    n = len(s)
    best =""
    for i in range(n):
        for j in range(i,n):
            sub = s[i:j+1]
            print(sub)
            if sub == sub[::-1] and len(sub) > len(best):
                best= sub
    return best
print(longest_palindrome_bf("abba"))
```
