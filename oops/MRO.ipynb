```python
class Human:
    def eat(self):
        print("I can eat")

class Male(Human):
    def eat(self):
        print("Male can eat")

m = Male()

#1, Hard-coded parent call

print(Human.eat(m))

#2, MRO aware, cleaner, preferred when multiple inheritance may be involved.
print(super(Male, m).eat())
```
