Design Patterns
│
├── Creational  → Object kaise CREATE hoga?
├── Structural  → Objects/classes kaise CONNECT honge?
└── Behavioral  → Objects kaise COMMUNICATE/BEHAVE karenge?


Creational
│
├── Factory Method   → Child decides object
├── Abstract Factory → Related objects ki family
├── Builder          → Complex object step-by-step
├── Prototype        → Existing object ko clone
└── Singleton        → Sirf ek instance

Structural Patterns
│
├── Adapter      → CONNECT
├── Bridge       → SEPARATE
├── Composite    → TREE
├── Decorator    → ADD
├── Facade       → SIMPLIFY
├── Flyweight    → SHARE
└── Proxy        → CONTROL

Behavioral
│
├── Chain of Responsibility     Request ko handlers ki chain mein pass karo
├── Command                     Request/action ko object bana do
├── Interpreter                 Language/expression ko interpret karo
├── Iterator                    Collection ko sequentially traverse karo
├── Mediator                    Objects ke communication ko centralobject handle kare
├── Memento                     Object ki previous state save/restore karo
├── Observer                    Ek object change ho → multiple objects ko notify karo
├── State                Object ka behavior uski current state ke according change ho
├── Strategy                    Runtime par algorithm/behavior choose karo
├── Template Method      Algorithm ka skeleton parent mein, specific steps child mein
└── Visitor              Existing objects ko modify kiye bina new operation add karo


```python
class Logger:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance
```

```python
"""
Factory:
To separate object creation from object usage, 
so the client doesn't need to know which concrete class needs to be instantiated
Parent → defines HOW to create
Child  → decides WHAT to create
"""
#"Kya mujhe runtime par different types ke objects create karne hain?"
class UPI:
  def pay(self,amt):
    print("pay using upi")

class Card:
  def pay(self,amt):
      print("pay using card")

class Cash:
  def pay(sel,amt):
    print ("pay using cash")


class PaymentFactory:
  @staticmethod
  def create_payment(payment_type):
    if payment_type=="UPI":
      return UPI()
    elif payment_type=="Card":
      return Card()
    elif payment_type =="Cash":
      return Cash()
    raise ValueError("unsurpported payment type")

payment = PaymentFactory.create_payment("UPI")
payment.pay(1000)
```

```python
"""
Builder Pattern ka use tab hota hai 
jab ek object ko banane ke liye bahut saare parameters/options ho, 
aur hum us object ko step-by-step construct karna chahte hain.
"""

class Computer:

    def __init__(self):
        self.cpu = None
        self.ram = None
        self.storage = None
        self.gpu = None
        self.os = None
        self.wifi = False
        self.bluetooth = False


class ComputerBuilder:

    def __init__(self):
        self.computer = Computer()

    def set_cpu(self, cpu):
        self.computer.cpu = cpu
        return self

    def set_ram(self, ram):
        self.computer.ram = ram
        return self

    def set_storage(self, storage):
        self.computer.storage = storage
        return self

    def set_gpu(self, gpu):
        self.computer.gpu = gpu
        return self

    def set_os(self, os):
        self.computer.os = os
        return self

    def enable_wifi(self):
        self.computer.wifi = True
        return self

    def enable_bluetooth(self):
        self.computer.bluetooth = True
        return self

    def build(self):
        return self.computer


computer1 = (ComputerBuilder().set_cpu("ultra").set_ram("64gb").build())

#print(computer1)
print(computer1.cpu)
print(computer1.ram)
```
