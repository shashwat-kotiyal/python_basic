```python
import datetime


class Employee:
    emp_count = 0
    raise_amt = 1.05

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.emp_count += 1

    def apply_raise(self):
        self.salary = int(self.raise_amt * self.salary)

    @classmethod
    def set_raise_amt(cls, amount):
        # Print("use to make alternate constructor")
        print("This is a class method")
        print(f"Accessing class_var: {cls.raise_amt}")
        cls.raise_amt = amount

    @classmethod
    def from_string(cls, emp_str):
        name, salary = emp_str.split("-")
        return cls(name, salary)

    @staticmethod
    def is_valid_raise(raise_amt):  # provide utility function
        return 1 <= raise_amt <= 1.05

    @staticmethod
    def is_weekday(day):  # provide utility function
        if day.weekday() == 5 or day.weekday() == 5:
            return False
        return True

    def __str__(self):
        return f"{self.name}-{self.salary}"

    def __repr__(self):
        return f"Employee({self.name!r},{self.salary})"

    def __len__(self):
        return len(self.name)


print(Employee.emp_count)
emp1 = Employee("prashant", 3600000)
emp2 = Employee("mohit", 3200000)

print(Employee.emp_count)
print(emp1.emp_count)
print(emp2.emp_count)


print(Employee.raise_amt)
print(emp1.raise_amt)
print(emp2.raise_amt)

print(Employee.is_valid_raise(0.4))
```

```python

class Developer(Employee):
    def __init__(self, name, pay, prg_lang):
        super().__init__(name, pay)
        # Employee.__init__(self,name,pay)   #both super ad this is expectable
        # super().__init__(name, pay) Continue method lookup according to Python's MRO (Method Resolution Order) different than hardcoding.
        #mein Python appropriate object (self) ko method binding ke through handle karta hai.
        #That's why self nahi likhte.
        self.prg_lang = prg_lang


dev2 = Developer("ranjit", 600000, "c++")


class Manager(Employee):
    def __init__(self, name, pay, employ=None):
        super().__init__(name, pay)
        # Employee.__init__(self,name,pay)   #both super ad this is expectable
        if employ is None:
            self.employ = []
        else:
            self.employee = employ

    def add_employee(self, emp):
        if emp not in self.employ:
            self.employ.append(emp)

    def remove_employee(self, emp):
        if emp in self.employ:
            self.employ.remove(emp)


mag1 = Manager("Mohit", 1000000, [emp1])

print("*" * 5 + "isinstance" + "*" * 5)
print(isinstance(mag1, Manager))
print(isinstance(mag1, Employee))
print(isinstance(mag1, Developer))
print(issubclass(Developer, Employee))
```

```python
class Employee:

    def __new__(cls, name):
        print("__new__ called")
        return super().__new__(cls)

    def __init__(self, name):
        print("__init__ called")
        self.name = name


e1 = Employee("Shashwat")

"""
Python internally roughly ye karta hai:
obj = Class.__new__(Class, ...)
Class.__init__(obj, ...)
__new__ class method jaisa behave karta hai because it receives cls,
but technically __new__ static method semantics ke saath special method hai.

__new__is used when you need to control the creation of a new instance 
while __init__is used when you need to control the initialization of a new 
instance
__new__is the first step of instance creation. It's called first and is 
responsible for returning a new instance of your class
__init__doesn't return anything; it's only responsible for initializing the 
instance after it's been created
"""
```
