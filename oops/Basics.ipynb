```python
#operator ploymorphism

print(10+20)
print("10"+"20")

#Method polymorphism
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")


#ducktyping
class Bird:
    def fly(self):
        print("Bird is flying")     

class Airplane:
    def fly(self):
        print("Airplane is flying")

#function overloading
def add(a, b):
    return a + b

def add(a, b, c=0):
    return a + b + c    

#function overriding
class Animal:

    def sound(self):
        print("Animal sound")


class Dog(Animal):

    def sound(self):
        print("Bark")


class Cat(Animal):

    def sound(self):
        print("Meow") 



#overloading 
def add(*args):
    return sum(args)

print(add(10, 20))
print(add(10, 20, 30))
```

single responsibilty(SRP)
A class should have only one responsibility, or more precisely, one reason to change

open to extension close for modification(OSP)
"Software entities should be open for extension but closed for modification."

Liskov substutuion(LSP)
Child class ko parent class ki jagah use karo, toh existing program ka expected behavior break nahi hona chahiye

interface segrretion
Dont force class to inherit method which is never used.
Client ko aise methods/interfaces par depend nahi karna chahiye jinki usse zarurat hi nahi hai.
Bada interface mat banao jisme sab kuch ho. Usko small, focused interfaces mein divide karo.

dependency inversion
High-level code ko low-level implementation par directly depend nahi karna chahiye. Dono abstraction par depend karein.

any concreate class should not depend on other concrete class should depend on interface
Dependency ko class ke andar create karne ke bajay bahar se provide karo.


```python
class Bird:
    pass


class FlyingBird(Bird):

    def fly(self):
        print("Flying")


class Sparrow(FlyingBird):

    def fly(self):
        print("Sparrow flying")


class Penguin(Bird):

    def swim(self):
        print("Penguin swimming")
```

```python
class Printer:

    def print(self):
        pass


class Scanner:

    def scan(self):
        pass


class Fax:

    def fax(self):
        pass

class SimplePrinter(Printer):
    def print(self):
        print("Printing")

class MultiFunctionPrinter(Printer, Scanner, Fax):

    def print(self):
        print("Printing")

    def scan(self):
        print("Scanning")

    def fax(self):
        print("Faxing")
```
