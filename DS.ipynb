```python
def divide(arr):
    num_freq = {}
    for ele in arr:
        if ele in num_freq:
            num_freq[ele] += 1
        else:
            num_freq[ele] = 1
    for k, v in num_freq.items():
        if v % 2 != 0:
            return False
    return True


arr = [3, 2, 3, 2, 2, 2]
print(divide(arr))


from collections import Counter


def divide1(arr):
    arr = Counter(arr)
    print(arr)
    for key in arr:
        arr[key] % 2 == 1
        return False
    return True


divide1(arr)
```

```python
# minimum no of operation to make uni grid with  operataion add and sub x
# grid = [[2,4], [6,8]] , x=2 , o/p =4
grid = [[2, 4], [6, 8]]
x = 2


def minOperations(grid, x):
    arr = [col for row in grid for col in row]
    print(arr)
    arr.sort()

    mid = arr[len(arr) // 2]  # median
    rem = arr[0] % x
    res = 0
    for d in arr:
        if d % x != rem:
            return -1
        res = res + abs(d - mid) // x
    return res


minOperations(grid, 4)
```
