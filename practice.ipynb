```python
def repeat(n):
    def deco(ori_func):
        def inner_func(*args, **kargs):
            for _ in range(n):
                ori_func(*args, *kargs)

        return inner_func

    return deco

@repeat(2)
def add(a, b):
    print(a + b)

add(2, 3)
```

```python
# run function n times
def repeat(n):
    def deco(original): 
        def inner(*args,**kargs):
            for _ in range(n):
                print(original(*args,**kargs))
        return inner
    return deco

@repeat(2)
def sum(a,b):
    return a+b

sum(3,4)
```

```python
# Create a decorator @time_it that measures how long a function takes to run.
from datetime import datetime

def time_taken(original):
    def inner(*args,**kargs):
        start_time= datetime.now()
        print(original(*args,**kargs))
        end_time = datetime.now()
        print(f'time_taken={end_time - start_time}')
    return inner

@time_taken
def wice(a,b):
    return(a+b)

wice(3,6)


```
