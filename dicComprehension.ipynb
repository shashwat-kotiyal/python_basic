```python
#Create a dictionary where the keys are numbers and the values are their squares:
nums = [1, 2, 3, 4]

d = { ele : ele **2 for ele in nums}
print(d)
#Create a dictionary where only even numbers are included, with values as cubes:
d = {ele:ele**3 for ele in nums if ele%2 == 0}
print(d)

fruit_prices = {'apple': 100, 'banana': 50, 'mango': 80}

swap_fruit = {value:key for key,value in fruit_prices.items()}
print(swap_fruit)
```

```python
#Generate a dictionary of multiplication tables from 1 to 3:
# Output:
# {
#   1: {1: 1, 2: 2, 3: 3},
#   2: {1: 2, 2: 4, 3: 6},
#   3: {1: 3, 2: 6, 3: 9}
# }

table = { ele: {ele*num for num in range(1,11)} for ele in range(1,4)}

print(table)
```

```python
#Create a dictionary of students who scored >= 50, increasing their marks by 5:
marks = {'Amit': 50, 'Neha': 80, 'Raj': 35, 'Simran': 90}

marks = {k:v+5 for k,v in marks.items() if v>35}
print(marks)
```
