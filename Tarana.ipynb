```python
#QUick sort
def quicksort(arr, count=0):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr)//2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x==pivot]
        right = [x for x in arr if x > pivot]
        count +=1
        # # Recursively sort left and right, accumulating the count
        # left, count = quicksort(left, count)
        # right, count = quicksort(right, count)

        print(f"pass{count}:{left ,middle,right}")
        return quicksort(left) + middle + quicksort(right)



arr = [64, 34, 25, 12, 22, 2 ,5, 11, 90]
sorted_arr = quicksort(arr)
print("Sorted array:", sorted_arr)
```
